# Cyclist Bike-Share Case Study - Data Cleaning Process

**Objective:** Prepare the four 2019 quarterly Cyclist trip files for comparing ride behaviour between casual riders and annual members.

## Workflow

1. Load and inspect the quarterly files.
2. Standardize column names and combine the data.
3. Keep the columns needed for the business question.
4. Standardize data types and customer labels.
5. Create `ride_length` and `day_of_week` from the timestamps.
6. Review invalid durations, inspect unusually long rides, and run final checks.
7. Export the cleaned dataset for the analysis notebook.


> **Before running:** The four `read_csv` lines below point to a folder on the original Windows computer. Keep the source CSV files at those paths, or update the paths for your own computer before running the notebook from top to bottom. The source files are not included in this notebook.

## 4. Combine the quarterly datasets

Concatenate the four quarterly DataFrames into one table, then reset the index so row labels run continuously.


```python
# merge the four datasets
data = pd.concat([q1, q2, q3, q4], axis=0)
```

```python
data.info() # see there are indices not aligned
```

```python
data = data.reset_index(drop=True) # reset indices
data.info()
```

## 5. Select columns for the analysis

The business question is about differences between casual riders and annual members. The `gender` and `birthyear` columns are not needed for this comparison, so they are removed from the working dataset. Their missing values do not require dropping otherwise usable trips.


```python
data = data.drop(['gender','birthyear'], axis=1)
data.info()
```

## 6. Standardize data types

Convert the timestamps to datetime values and prepare the supplied `tripduration` field as a numeric duration in seconds.

### Convert timestamps

Parse `start_time` and `end_time` as datetime values.


```python
# convert start_time and end_time from 'str' to 'datetime' datatype
data['start_time'] = pd.to_datetime(data['start_time'])
data['end_time'] = pd.to_datetime(data['end_time'])
data.info()
```

### Inspect the combined data

Review sample records and the supplied duration values before converting `tripduration` to a numeric type.


```python
data.head()
```

```python
data['tripduration'].head()
```

```python
pd.to_numeric(data['tripduration'], errors='coerce').head()
```

### Prepare `tripduration` for numeric conversion

The duration values contain commas as thousands separators. Remove those commas before converting the column to numeric values.


```python
data['tripduration'].head()
```

```python
data['tripduration'].str # this let's us access string methods and apply to column which contains strings
```

```python
# this will remove all the commas in our tripduration column
data['tripduration'].str.replace(",","").head() 
```

```python
data['tripduration'] = pd.to_numeric(data['tripduration'].str.replace(',',''), errors='raise')
```

```python
data['tripduration'].head()
```

```python
data['tripduration'].info()
```

```python
data['tripduration'].dtype
```

```python
data['tripduration'].isnull().sum()
```

### Diagnostic check: duration calculated from timestamps

Temporarily calculate elapsed seconds from `start_time` and `end_time` to inspect the records. This first derived column is removed after the check; the final `ride_length` field is created later alongside `day_of_week`.


```python
# now let's calculate our ride_length, the calculated time from start_time and end_time in seconds

data['ride_length'] = (data['end_time']-data['start_time']).dt.total_seconds() # we write .dt to access time methods
```

```python
data['ride_length'].info()
```

```python
data['ride_length'].describe()
```

```python
data.info()
```

```python
data = data.drop('ride_length', axis=1)
```

```python
data.info()
```

## 7. Standardize customer types

Inspect the `usertype` column, check for missing labels, and standardize the two categories to `Member` and `Casual` for the analysis.


```python
data['usertype'].info()
```

```python
data['usertype'].isnull().sum()
```

## 8. Validate the supplied duration

Review the distribution of the original `tripduration` field and count non-positive values. Keep this separate from the timestamp-derived `ride_length` check so the two duration measures are not confused.


```python
# validate tripduration
print(data["tripduration"].describe())
```

```python
# any negative values in tripduration
print((data['tripduration'] <= 0).sum()) 
```

```python
data.info()
```

## 9. Create the analysis columns

Create two fields from the timestamps:

- `ride_length`: elapsed time from `start_time` to `end_time`, stored in seconds.
- `day_of_week`: the weekday on which the ride started, derived from `start_time`.

These will be used in the member-versus-casual analysis. The original `tripduration` column is retained so its values can be compared with the calculated duration.


```python
# let's get back to cleaning
```

```python
data.info()
```

```python
data['ride_length'] = data['end_time']-data['start_time']
data['ride_length'].info()
```

```python
data['ride_length'] = data['ride_length'].dt.total_seconds()
```

```python
data['ride_length'].info()
```

```python
data['ride_length'].head()
```

```python
data['ride_length'] = data['ride_length'].astype(int)
```

```python
data['ride_length'].head()
```

```python
data.info()
```

```python
# now calculate week of the day column
data['day_of_week'] = data['start_time'].dt.dayofweek # 1 for monday, 6 for sunday
```

**Weekday numbering note:** pandas `.dt.dayofweek` uses `0 = Monday` through `6 = Sunday`. Use this mapping when interpreting `day_of_week`; weekday labels can be added later for presentation.


```python
data['day_of_week'].sample(5)
```

```python
data['day_of_week'].describe()
```

### Check non-positive calculated durations

Count records with `ride_length` at or below zero. These records cannot represent a positive elapsed ride duration, so the next cell keeps only rows with `ride_length > 0`.


```python
data[data['ride_length'] <= 0].count()['trip_id'] # negative or zero distance rides
```

```python
data.info()
```

```python
data = data[data['ride_length'] > 0] # 13 rows removed.
```

```python
data.info()
```

## 10. Review unusually long durations

Inspect rides longer than 24 hours using the supplied `tripduration` field, then compare the supplied duration with calculated `ride_length` for the 50 longest calculated rides. This is a review step, not an automatic outlier-removal rule: very long values are kept unless there is evidence they are invalid.


```python
# Number of rides lasting more than 24 hours
long_rides = data[data["tripduration"] > 86400]

print("Rides over 24 hours:", len(long_rides))

print(
    "Percentage of all rides:",
    round(len(long_rides) / len(data) * 100, 3),
    "%"
)

# Inspect the 10 longest rides
display(
    long_rides.sort_values(
        "tripduration",
        ascending=False
    )[
        [
            "trip_id",
            "start_time",
            "end_time",
            "tripduration",
            "usertype"
        ]
    ].head(10)
)
```

### Compare the two duration fields for the longest calculated rides

The difference below is `tripduration - ride_length`, in seconds. Small differences can arise from how the source fields were recorded; a difference by itself is not enough to label a trip faulty. Review the timestamps and source context before making any exclusion decision.


```python
temp = data.sort_values('ride_length', ascending=False).head(50)[['trip_id','ride_length','tripduration']]
temp.head()
```

```python
temp['tripduration']-temp['ride_length'] 
```
